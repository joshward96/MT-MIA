```python
import pandas as pd

# Read the CSV files
activity = pd.read_csv('Customer Flight Activity.csv')
loyalty_history = pd.read_csv('Customer Loyalty History.csv')

# 1. Change 'Loyalty Number' to 'loyalty_history_id' in loyalty_history
if 'Loyalty Number' in loyalty_history.columns:
    loyalty_history.rename(columns={'Loyalty Number': 'loyalty_history_id'}, inplace=True)

# 2. Add 'activity_id' as the first column in activity
activity.insert(0, 'activity_id', range(1, len(activity) + 1))

# 2b. Rename 'Loyalty Number' to 'loyalty_history_id' in activity (before snake_case conversion)
if 'Loyalty Number' in activity.columns:
    activity.rename(columns={'Loyalty Number': 'loyalty_history_id'}, inplace=True)

# 3. Add underscores to all column names in both dataframes
# Function to convert column names to snake_case
def to_snake_case(name):
    # Replace spaces with underscores and convert to lowercase
    return name.replace(' ', '_').lower()

# Apply to activity dataframe
activity.columns = [to_snake_case(col) for col in activity.columns]

# Move loyalty_history_id to be the last column in activity
if 'loyalty_history_id' in activity.columns:
    # Get all columns except loyalty_history_id
    cols = [col for col in activity.columns if col != 'loyalty_history_id']
    # Add loyalty_history_id at the end
    cols.append('loyalty_history_id')
    # Reorder the dataframe
    activity = activity[cols]

# Apply to loyalty_history dataframe
loyalty_history.columns = [to_snake_case(col) for col in loyalty_history.columns]

# Display the results
print("Activity DataFrame:")
print(activity.head())
print(f"\nActivity columns: {list(activity.columns)}")
print(f"\nActivity shape: {activity.shape}")

print("\n" + "="*80 + "\n")

print("Loyalty History DataFrame:")
print(loyalty_history.head())
print(f"\nLoyalty History columns: {list(loyalty_history.columns)}")
print(f"\nLoyalty History shape: {loyalty_history.shape}")

# Save the modified dataframes
activity.to_csv('cleaned/activity.csv', index=False)
loyalty_history.to_csv('cleaned/loyalty_history.csv', index=False)
```
